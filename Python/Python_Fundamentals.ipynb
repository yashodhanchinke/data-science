### Conditional  

- Conditional statements allow a program to make dicisions based on coditions. They execute different blocks of code depending on whether a codition True or False

#### 1) If-statement

- Executes a block only if codintion is True

````python
""" syntax 
if codition:
     code """
````

````python
age = int(input("Enter Your Age"))

if age > 18:
    print("You are Eligible for voting")
    
````

#### 2) if-else Statement

- Execute one block if the condition is True; otherwise executes another block.

````python
n1 = int(input("Enter the first number"))
n2 = int(input("Enter the second number"))

if n1 > n2 :
    print(f"{n1} is greater than {n2} ")
else :
    print(f"{n2} is greater than {n1} ")
````

#### 3) if-elif-else statement

- Used to check multiple conditions

````python
age = int(input("Enter Your Age"))

if age >= 18 :
    print("You are Eligible for Vote")
elif age <= 0 :
    print("Invalid Input")
else:
    print("You are not eligible for the vote")
````

````python
marks = int(input("Enter Your Marks"))

if marks >= 75 :
    print("Distinction")
elif 60 <= marks < 75:
    print("first class with distction")
elif 40 <= marks <60 :
    print("second class with distiction")
elif marks < 35:
    print("Fail")
````

###  Variables in Python

A **variable** is a named memory location used to store data. It acts as a container that holds values, which can be changed during program execution.

#### Syntax

```python
variable_name = value
```

#### Example

```python
name = "Mayur"
age = 25
salary = 50000
is_student = False

print(name)
print(age)
print(salary)
print(is_student)
```

### Rules for Naming Variables

#### Valid Rules

- Variable names can contain **letters (A-Z, a-z), digits (0-9), and underscores (_)**.
- A variable name **must start with a letter or an underscore (_)**.
- Variable names are **case-sensitive** (`age`, `Age`, and `AGE` are different variables).
- Use **meaningful names** for better readability.

##### Valid Examples

```python
name = "John"
_age = 25
student1 = "Rahul"
total_marks = 450
price2025 = 199.99
```

##### Invalid Variable Names

- Cannot start with a number.
- Cannot contain spaces.
- Cannot use special characters like `@`, `#`, `$`, `%`, `&`.
- Cannot use Python keywords (`if`, `for`, `while`, `class`, etc.).

### Difference Between Parameter and Argument

| **Parameter** | **Argument** |
|---|---|
| Variable defined in the function definition | Actual value passed to the function |

| Parameters exist inside the function definition | Arguments exist in the function call |

- Parameter = a variable/placeholder written in the function definition.

- Parameter = receives the value
- Argument = gives the value

| `return` | `print()` |
|---|---|
| Sends a value back from a function. | Displays a value on the screen/console. |

| The returned value can be stored in a variable. | The displayed value cannot be directly stored as the output of `print()`. |
| Ends the function execution. | Does not end the function execution. |
| Mainly used when the result needs to be used later. | Mainly used when you want to show information to the user. |
| `return` is a Python keyword. | `print()` is a built-in Python function. |
| Example: `return result` | Example: `print(result)` |

- return → Gives a value back from a function.

- In positional arguments, values are assigned to parameters based on their position/order

````python
def student(name,age,course):
    print(f"Name : {name}")
    print(f"Age : {age}")
    print(f"Course : {course}")

student("Rahul",22,"Data Science")
````

#### Python `*args` and `**kwargs`

#### What is `*args`?

- The `*args` parameter allows a function to accept any number of positional arguments.
- Inside the function, `args` becomes a tuple containing all the passed arguments.


````python
def rank(*args):
    return args[0] + args[1]
````

````python
rank("Sachine"," Harshal","Pallavi","Harsh","Sanchita")
````

#### Keywords Arguments
- In keyword arguments, values are assigned to parameters based on their parameter names.

````python
def rank(first,second,third):
    return first,second
rank(second="Harshal",first="Sachine",third="Pallavi")
````

#### Arbitrary Keyword Arguments - `**kwargs`

- If you do not know how many keyword arguments will be passed into your function, add two asterisks `**` before the parameter name.
- This way, the function will receive a dictionary of arguments and can access the items accordingly.

````python
def rank(**kwargs):
    return kwargs
rank(name="yash",age=25,course="data science")
````

### What is Local and Global Variable

#### Local Variable
- Local variables are defined inside a function and exist only during its execution.
- They cannot be accessed from outside the function.

````python
def display():
    x=10 #local variable
    print(x)
display()
````

#### Global Variable

- A global variable is a variable that is created outside a function. It can be accessed from anywhere in the program.

````python
x = 10 #global variable
def display():
    print(x)
display()
````

#### Global Keyword

- The `global` keyword is used when we want to modify a global variable inside a function.

````python
x = 10
def change():
    global x
    x = 20

change()
print(x)
````

### Higher-Order Functions

- A higher-order function takes another function as an argument, returns one, or both.
- They are commonly used with Python's functional-style tools.

#### `map()`

- Applies a given function to all items in an input list and returns a map object (an iterator).

````python
num = [2,4,5,8,7,9,5,4,2,6,1]
var = list(map(lambda x : x**2,num))
var
````

````python
n1 = [1,2,3,4,5]
n2 = [6,7,8,9,10]
Sum = list(map(lambda a,b:a+b,n1,n2))
Sum
````

- The `filter()` function is used to filter elements from an iterable based on a condition.
